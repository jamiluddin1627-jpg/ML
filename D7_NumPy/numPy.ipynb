Array Creation

```python
import numpy as np
```

```python
# 1D Array creation
a = [1, 2, 3, 4]
arr = np.array(a)
print(arr)
print(type(arr))
```

```python
# 2D Array
a = [1 , 2, 3, 4]
b = [5, 6, 7, 8]
ar1 = np.array([a, b])
print(ar1)
print(type(ar1))
```
