Array Creation

```python
import numpy as np
```

```python
# 1D Array creation
a = [1, 2, 3, 4]
arr = np.array(a)
print(arr)
print(type(arr))
```

```python
# 2D Array
a = [1 , 2, 3, 4]
b = [5, 6, 7, 8]
ar2 = np.array([a, b])
print(ar2)
print(type(ar2))
```

```python
# 3D Array
ar3 = np.array([[a, b], [b, a]])
print(ar3)
print(type(ar3))
```

Initial Placeholder

```python
# arange , ones, zeros, random
a = np.arange(5) # 1D
print(a)

b = np.arange(6).reshape(3, 2) # 2D
print(b)

c = np.arange(12).reshape(2, 2, 3) # 3D

print(c)
```
